```python
products = pd.read_csv("https://raw.githubusercontent.com/MohammedHameds/test1455/refs/heads/main/products.csv")

products
```

```python
products.drop_duplicates()
```

```python
products['category'] = products['category'].str.strip().str.title()
products['product_name'] = products['product_name'].str.strip().str.title()
```

```python
products.dtypes
```

```python
products['unit_price'] = pd.to_numeric(products['unit_price'], errors='coerce')
```

```python
orders = orders.drop_duplicates(subset=['order_id'])
```

```python
sales = pd.merge(orders, customers, on='customer_id', how='left')
sales = pd.merge(sales, products, on='product_id', how='left')
```

```python
sales
```

```python
sales['total_amount'] = sales['quantity'] * sales['unit_price']
sales = sales.dropna(subset=['full_name'])
sales_final = sales[['order_id', 'full_name', 'product_name', 'quantity', 'unit_price', 'total_amount']]
sales_final = sales_final.rename(columns={'full_name': 'customer', 'product_name': 'product'})
```

```python
sales_final
```

```python
sales_final.to_csv("sales_final")
```

```python
import pyodbc

conn = pyodbc.connect(
    "driver={ODBC Driver 17 for SQL Server};"
    "server=.;"
    "database=sales_final;"
    "trusted_connection=yes;"
)
```

```python
cursor = conn.cursor()

cursor.execute("""
    IF NOT EXISTS (SELECT * FROM sysobjects WHERE name='Sales')
    CREATE TABLE Sales (
        order_id INT,
        customer NVARCHAR(100),
        product NVARCHAR(100),
        quantity INT,
        unit_price FLOAT,
        total_amount FLOAT
    )
""")
conn.commit()
```

```python
for index, row in sales_final.iterrows():
    cursor.execute(
        "INSERT INTO Sales (order_id, customer, product, quantity, unit_price, total_amount) VALUES (?, ?, ?, ?, ?, ?)",
        row['order_id'], row['customer'], row['product'], row['quantity'], row['unit_price'], row['total_amount']
    )

conn.commit()
```

```python
from flask import Flask, render_template
import pyodbc
import pandas as pd

app = Flask(__name__)

def get_sales_data():
    conn = pyodbc.connect(
        "driver={ODBC Driver 17 for SQL Server};"
        "server=.;"
        "database=sales_final;"
        "trusted_connection=yes;"
    )
    df = pd.read_sql("SELECT * FROM Sales", conn)
    conn.close()
    return df

@app.route('/')
def dashboard():
    try:
        df = get_sales_data()
        return render_template(
            'dashboard.html',
            data=df.to_dict(orient='records'),
            columns=df.columns.tolist()
        )
    except Exception as e:
        return f"<h2>An error occurred:</h2><pre>{e}</pre>"

if __name__ == '__main__':
    app.run(debug=True, use_reloader=False)
```
